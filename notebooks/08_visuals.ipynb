```python
import pandas as pd
import plotly.express as px
import matplotlib.pyplot as plt

df_final = pd.read_csv("../data/processed/happiness_projections_2011_2030.csv")
```

```python
df_final.columns
```

```python
df_final.head(12)

```

```python
country = "United States"
df_country = df_final[df_final["country_name"] == country]

df_country = df_country.sort_values("year").reset_index(drop=True)
df_country["life_evaluation_3_year_average"] = (
    df_country["life_evaluation_3_year_average"].interpolate()
)

fig = px.line(
    df_country,
    x="year",
    y="life_evaluation_3_year_average",
    title=f"Évolution du bonheur – {country} (2014–2030)",
    markers=True
)

fig.update_traces(connectgaps=True)
fig.show()
```

```python
country = "Belgium"
df_country = df_final[df_final["country_name"] == country].sort_values("year")

fig = px.line(
    df_country,
    x="year",
    y="life_evaluation_3_year_average",
    title=f"Évolution du bonheur – {country} (2011–2030)",
    markers=True
)

fig.update_traces(connectgaps=True)
fig.show()
```

# Top 10 - Happiness en 2030

```python
df_2030 = df_final[df_final["year"] == 2030].copy()
```

```python
top10 = df_2030.nlargest(10, "life_evaluation_3_year_average")
```

```python
fig = px.bar(
    top10.sort_values("life_evaluation_3_year_average"),
    x="life_evaluation_3_year_average",
    y="country_name",
    orientation="h",
    title="Top 10 des pays les plus heureux en 2030",
    color="life_evaluation_3_year_average",
    color_continuous_scale="Blues"
)
fig.show()
```

# Flop 10

```python
flop10 = df_2030.nsmallest(10, "life_evaluation_3_year_average")
```

```python
fig = px.bar(
    flop10.sort_values("life_evaluation_3_year_average"),
    x="life_evaluation_3_year_average",
    y="country_name",
    orientation="h",
    title="Flop 10 des pays les moins heureux en 2030",
    color="life_evaluation_3_year_average",
    color_continuous_scale="Reds"
)
fig.show()
```

# Carte Mondiale

```python
fig = px.choropleth(
    df_2030,
    locations="country_name",
    locationmode="country names",
    color="life_evaluation_3_year_average",
    color_continuous_scale="Viridis",
    title="Bonheur mondial en 2030",
)
fig.update_layout(
    width=1400,
    height=700,
    geo=dict(showframe=False, showcoastlines=False)
)


fig.show()
```

```python
df_2030 = df_final[df_final["year"] == 2030].copy()

fig = px.choropleth(
    df_2030,
    locations="country_name",
    locationmode="country names",
    color="life_evaluation_3_year_average",
    hover_name="country_name",
    hover_data={
        "life_evaluation_3_year_average": True,
        "year": False
    },
    color_continuous_scale="RdYlGn",
    title="Bonheur mondial en 2030",
)

fig.update_layout(
    geo=dict(showframe=False, showcoastlines=False),
    coloraxis_colorbar=dict(
        title="Bonheur",
        ticks="outside"
    )
)
fig.update_layout(
    autosize=True,
    height=700,
    margin=dict(l=0, r=0, t=50, b=0)
)


fig.show()
```

# animation

```python
df_final = df_final.sort_values("year")
df_final["year"] = df_final["year"].astype(int)

fig = px.choropleth(
    df_final,
    locations="country_name",
    locationmode="country names",
    color="life_evaluation_3_year_average",
    animation_frame="year",
    color_continuous_scale="RdYlGn",
    title="Évolution du bonheur mondial (2011–2030)",
)

fig.update_layout(
    autosize=True,
    height=750,
    margin=dict(l=0, r=0, t=50, b=0),
    geo=dict(showframe=False, showcoastlines=False)
)

fig.show()
```

# Courbe multi-pays


```python
countries = [
    "Afghanistan",
    "Bolivia",
    "Botswana",
    "Bulgaria",
    "Costa Rica",
    "Finland",
    "United States",
    "China"
]

df_sel = df_final[df_final["country_name"].isin(countries)]

fig = px.line(
    df_sel,
    x="year",
    y="life_evaluation_3_year_average",
    color="country_name",
    markers=True,
    title="Évolution du bonheur (2011–2030) — 8 pays sélectionnés"
)

fig.update_traces(connectgaps=True)

fig.update_layout(
    autosize=True,
    height=650,
    margin=dict(l=0, r=0, t=50, b=0),
    legend_title_text="Pays"
)

fig.show()
```


### Lecture générale du graphique

Le modèle projette le niveau de bonheur en prolongeant les tendances historiques, en intégrant les indicateurs macro‑économiques et en appliquant un lissage temporel. Il ne prédit pas des événements politiques, mais extrapole les dynamiques observées dans les données réelles.

### Analyse rapide par pays (1 ligne chacun)

- **Afghanistan** : forte chute liée à la situation politique récente, suivie d’un léger rebond statistique dû à la stabilisation des indicateurs macro.  
- **Bolivia** : trajectoire globalement stable avec une légère tendance à la hausse, cohérente avec une croissance modérée et des indicateurs sociaux relativement constants.  
- **Botswana** : baisse progressive depuis 2014–2015, en ligne avec le ralentissement économique et les inégalités persistantes. (Chute du marché du diamant) 
- **Bulgaria** : amélioration régulière, portée par la croissance économique et l’intégration européenne.  
- **Costa Rica** : niveau élevé mais légèrement décroissant, reflétant des pressions économiques malgré un fort capital social.  
- **Finland** : stabilité à un niveau très élevé, cohérente avec ses performances sociales et institutionnelles.  
- **United States** : tendance légèrement descendante, influencée par les inégalités et les tensions sociales.  
- **China** : progression lente mais continue, portée par l’amélioration du niveau de vie malgré des défis sociaux.
