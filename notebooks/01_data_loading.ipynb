# Harmoniser les colonnes

Dans WHR :
Country name → majuscule + espace

Dans World Bank :
country_name → minuscule + underscore


```python
whr = whr.rename(columns={"Country name": "Country"})
wb = wb.rename(columns={"country_name": "Country"})
whr = whr.rename(columns={"Year": "year"})
```

```python
# verification

set(whr["Country"]) - set(wb["Country"])
```

```python
set(wb["Country"]) - set(whr["Country"])
```

```python
# Dictionnaire de correspondance

country_mapping = {
    "Congo": "Congo, Rep.",
    "DR Congo": "Congo, Dem. Rep.",
    "Côte d’Ivoire": "Cote d'Ivoire",
    "Egypt": "Egypt, Arab Rep.",
    "Gambia": "Gambia, The",
    "Hong Kong SAR of China": "Hong Kong SAR, China",
    "Iran": "Iran, Islamic Rep.",
    "Kyrgyzstan": "Kyrgyz Republic",
    "Republic of Korea": "Korea, Rep.",
    "Republic of Moldova": "Moldova",
    "Slovakia": "Slovak Republic",
    "Swaziland": "Eswatini",
    "Venezuela": "Venezuela, RB",
    "Türkiye": "Turkiye",
    "Yemen": "Yemen, Rep.",
    "Taiwan Province of China": None,  # pas dans WB
    "Somaliland Region": None,         # pas dans WB
    "State of Palestine": None,        # pas dans WB
    "North Cyprus": None,              # pas dans WB
    "Syria": None                      # pas dans WB (données manquantes)
}
```

```python
whr["Country"] = whr["Country"].replace(country_mapping)
```

```python
whr["Country"] = whr["Country"].replace({
    "Turkey": "Turkiye",
    "Türkiye": "Turkiye"
})
```

```python
# Vérifier à nouveau les différences

set(whr["Country"]) - set(wb["Country"])
```

# Merge de 2 tableaux

```python
merged = pd.merge(
    whr,
    wb,
    on=["Country", "year"],
    how="inner"
)
```

```python
merged["Country"] = merged["Country"].replace({"Turkiye": "Turkey"})
```
