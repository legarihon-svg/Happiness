```python
wb = pd.read_csv('/content/drive/MyDrive/Jedha-Fullstack/FS_ProjetFinal_Room3/world_bank_data_2025.csv')
wb.head()
```

```python
print(whr.columns)
print(wb.columns)
```

```python
# on met les colonnes des 2 datasets en snake_case pour simplifier la suite du process

def clean_columns(df):
    df.columns = (
        df.columns
        .str.strip()                 # enlève espaces début/fin
        .str.lower()                 # minuscule
        .str.replace(' ', '_')       # espace → underscore
        .str.replace('-', '_')       # tiret → underscore
        .str.replace(r'[^\w]', '', regex=True)  # enlève caractères spéciaux
    )
    return df

whr = clean_columns(whr)
wb  = clean_columns(wb)
```

```python
print(whr.columns)
print(wb.columns)
```

```python
# On crée les colonnes iso 3

whr['iso_3'] = coco.convert(names=whr['country_name'], to='ISO3')
wb['iso_3'] = coco.convert(names=wb['country_name'], to='ISO3')

# Pas de conversion possible pour Channel Island !!! (ni Channel N°5)
```

```python
# on supprime les colonnes Channel Island de "wb" car elles ne se trouveny pas dans le dataset "whr"
# Elles sont donc inutiles

wb = wb[wb['country_name'] != 'Channel Islands'].copy()
```

```python
wb['iso3'] = coco.convert(names=wb['country_name'], to='ISO3')

```

```python
wb.head()
```

```python
whr.head()
```

```python
# création d'une colonne iso3 + year pour chaque datasets. On les utilisera comme primary key

whr['year'] = whr['year'].astype(str)
whr['iso3_year'] = whr['iso_3'] + whr['year']

wb['year'] = wb['year'].astype(str)
wb['iso3_year'] = wb['iso_3'] + wb['year']
```

```python
# checks que les clés sont vraiment uniques pour le join (afin d'éviter les doublons au join)

print(f"nombre de duplications de valeur pour whr = {whr.duplicated('iso3_year').sum()}")
print(f"nombre de duplications de valeur pour wb = {wb.duplicated('iso3_year').sum()}")
```

```python
# voir les lignes dupliquées

whr[whr.duplicated('iso3_year', keep=False)].sort_values('iso3_year')
```

```python
# on va vérifier si ces clés (de whr) existe dans wb autrement on les supprimera

whr[~whr['iso3_year'].isin(wb['iso3_year'])]['iso3_year'].unique()

# on voit que Taiwan ne se trouve pas dans WB
# on peut aussi déduire que "Cyprus" et "Somalia" existent mais pas "North Cyprus" ni "Somaliland Region"
# On peut donc supprimer ces lignes
```

```python
# suppresion des lignes 'North Cyprus' et 'Somaliland Region'

whr = whr[
    (whr['country_name'] != 'North Cyprus') &
    (whr['country_name'] != 'Somaliland Region')
].copy()

```

```python
# Ultime petit check


print(f"nombre de duplications de valeur pour whr = {whr.duplicated('iso3_year').sum()}")
print(f"nombre de duplications de valeur pour wb = {wb.duplicated('iso3_year').sum()}")
```

```python
# au final combien de clés communes avons nous et quelle pourcentage de perte

same_keys = len(set(whr['iso3_year']) & set(wb['iso3_year']))
print(f"nombre de clés communes = {same_keys}")

len_keys_whr = whr['iso3_year'].nunique()
print(f"nombre de clés WHR = {len_keys_whr}")

len_keys_wb = wb['iso3_year'].nunique()
print(f"nombre de clés WB = {len_keys_wb}")
```

```python
# pourcentage de perte dans chaque dataset après jointure

print(f"pourcentage de perte pour WHR = {round(100 - (same_keys / len_keys_whr * 100), 2)}%")
print(f"pourcentage de perte pour WB = {round(100 - (same_keys / len_keys_wb * 100), 2)}%")
```

```python
# merge par "iso3_year" comme primary key

df_final = pd.merge(
    whr,
    wb,
    on='iso3_year',
    how='inner'   # garde seulement les clés communes
)
```
