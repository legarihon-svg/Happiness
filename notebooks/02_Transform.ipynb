```python
df = df.astype({col: "string" for col in df.select_dtypes(include="object").columns})
```

```python
nulls_by_country = (
    df.isna()
      .groupby(df["country_name"])
      .sum()
)

nulls_by_country
```

```python
nulls_by_year = (
    df.isna()
      .groupby(df["year"])
      .sum()
)

nulls_by_year
```
